##### Working Machine probability matrix

```python
import numpy as np
import math

p_matrix = np.zeros(shape=[6,6])
def p(n,k):
    return math.comb(n,k)*(1/10)**k*(9/10)**(n-k)

for i in range(6):
    print(f'Number of working: {i}')
    for j in range(i+1):
        print(f'    Probability that {j} break: {round(p(i,j),4)}')
        p_matrix[i,j] = p(i,i-j)
np.set_printoptions(precision=3,suppress=True)
print(p_matrix)


            
```

```python
# Expected time till 5 break

p_break = 1/10
p_work = 9/10
n_machines = 5

# Iterating to create probability matrix T
T = np.zeros((6, 6))
for i in range(6):
    for j in range(i+1):
        k = i - j  # number that break
        T[i][j] = (np.math.comb(i, k) * (p_break**k) * (p_work**j))

Q = T[1:, 1:]  # transitions among states 1-5 only

# Creating linear system to avoid weird recurrence calculations
ones = np.ones(5)
A = np.eye(5) - Q

m = np.linalg.solve(A, ones)

for i, val in enumerate(m, start=1):
    print(f"Expected time from state {i}: {val:.4f} weeks")
```

```python
# Probabilty of being in state 1 at some random time in the future
import numpy as np
evals,evecs = np.linalg.eig(p_matrix)

evecs = [evecs[:,i] for i in range(len(evecs))]
print(evecs[-1])
print(p_matrix@evecs[-1])
```
