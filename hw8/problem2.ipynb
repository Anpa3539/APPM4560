### Problem 2: Receptor Bonding Markov Chain

```python
import numpy as np
from sympy import symbols, Matrix, linsolve

Q = np.array([[-2,1,1],
              [1,-1,0],
              [1,0,-1]])

eigenvalues,eigenvectors = np.linalg.eig(Q)
vecs = []
for i in range(len(eigenvalues)):
    val = eigenvalues[i]
    vec = eigenvectors[:, i] 
    vecs.append(vec)
    print(f"Eigenvalue {i+1}: {val}")
    print(f"Corresponding Eigenvector: {vec}\n")



```
