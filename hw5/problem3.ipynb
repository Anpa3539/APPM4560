### Stationary Distribution of Markov Chain

```python
import numpy as np
import sympy as sp

# Creating symbolic matrix
a = sp.symbols('a')
p = sp.Matrix([[1-a,a,0],[a,0,1-a],[0,1-a,a]])
A = p.T

eigenvalues_dict = A.eigenvals()
eigenvalues_list = list(eigenvalues_dict.keys())
# print(eigenvalues_list)

eigen_info = A.eigenvects()
for val, mult, vecs in eigen_info:
    print(f"Eigenvalue: {val} (Multiplicity: {mult})")
    print(f"Eigenvectors: {vecs}")
```

```python
# Evaluating the symbolic expressions when a=0.99 
a0 = 0.99
p = np.array([[1-a0,a0,0],[a0,0,1-a0],[0,1-a0,a0]])
A = p.T

evals,evecs = np.linalg.eig(A)

q0 = np.array([1,0,0])
coefficients = np.linalg.solve(evecs,q0)
print(coefficients)
```

```python
# Simulating the Markov Chain
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
def simulate_markov_chain(transition_matrix, start_state, steps):
    n_states = transition_matrix.shape[0]
    states = [start_state]
    current_state = start_state
    t1,t2,t3 = 0,0,0
    t1 +=1 # Initial value is 1 in this problem
    for _ in range(steps):
        # Sample next state based on current row of P
        next_state = np.random.choice(
            range(n_states), 
            p=transition_matrix[current_state]
        )
        states.append(next_state)
        current_state = next_state

        # Finding the time in each state
        if current_state == 0:
            t1 +=1
        elif current_state == 1:
            t2 +=1
        elif current_state == 2:
            t3 +=1
    time_in_states = [t1,t2,t3]
    return states,time_in_states

def fn(path,n):
    """ 
    Return the fraction of chains in state 1
    """
    count = 0
    for i in range(n):
        if path[i] == 0:
            count +=1
    f = count/(n+1)
    return f

# Givens 
a0 = 0.99
p = np.array([[1-a0,a0,0],[a0,0,1-a0],[0,1-a0,a0]])
A = p.T
N = [100,1000,10000]

# Looping through N values
for ni in N:
    path,time_in_states = simulate_markov_chain(p,0,ni)
    print(time_in_states)

    # Finding fraction time in state 1
    fraction = [fn(path,n) for n in range(ni)]

    fig = go.Figure()
    f_trace = go.Scatter(y=fraction,mode='lines',name='fn')
    fig.add_trace(f_trace)
    fig.add_hline(y=1/3, line_dash="dash", line_color="red", annotation_text="Stationary Distribution")
    fig.update_layout(title='Fraction of Time in State 1',xaxis_title='n',yaxis_title='fn')
    fig.show()
```
