### Simulating stationary distribution

```python
import numpy as np


def pn(n,K,a):
    return K*np.exp(n*a)
def qn(n,K,b):
    return K*np.exp(b*(n-1))

def pfunc(n,m,K,a,b):
    
    p = pn(n,K,a)
    q = qn(n,K,b)
    
    if n == 1:
        q = 0
    if n == 5:
        p = 0

    if n == m:
        return 1 - p - q
    elif m == n + 1:
        return p
    elif m == n - 1:
        return q
    else:
        return 0
    
p = np.zeros([5,5])
a = 0.04
b = 0.16
K = 0.1

for i in range(5):
    for j in range(5):
        p[i,j] = pfunc(i+1,j+1,K,a,b)
    print(sum(p[i]))

A = p.T

evals,evecs = np.linalg.eig(A)
evecs = [evecs[:,i] for i in range(len(evecs))]
sd = evecs[3]/sum(evecs[3])
print(sd)
```

### Simulating the Markov Chain

```python
def simulate_markov_chain(transition_matrix, start_state, steps):
    n_states = transition_matrix.shape[0]
    states = [start_state]
    current_state = start_state
    t1,t2,t3,t4,t5 = 0,0,0,0,0
    
    for _ in range(steps):
        # Sample next state based on current row of P
        next_state = np.random.choice(
            range(n_states), 
            p=transition_matrix[current_state]
        )
        states.append(next_state)
        current_state = next_state

        # Finding the time in each state
        if current_state == 0:
            t1 +=1
        elif current_state == 1:
            t2 +=1
        elif current_state == 2:
            t3 +=1
        elif current_state == 3:
            t4 +=1
        elif current_state == 4:
            t5 +=1
    time_in_states = {'t1':t1,'t2':t2,'t3':t3,'t4':t4,'t5':t5}
    return states,time_in_states

path_history,times = simulate_markov_chain(p,1,int(1e6))
print(times)


```

```python
import plotly.express as px
import plotly.graph_objects as go

time_in_states = [times['t1'],times['t2'],times['t3'],times['t4'],times['t5']]
print([t/1e6 for t in time_in_states])
x = [1,2,3,4,5]
fig = px.histogram(
    x=x,
    y=time_in_states,
    histnorm='probability density',
    nbins=5,
    category_orders=dict(day=["1","2","3","4","5"]),
    labels={'x':'State'}
)
fig.add_trace(go.Scatter(x=x, y=sd, mode='lines', name=''))

fig.update_yaxes(title='Percentage of Time in State')

fig.update_layout(bargap=0.2)
fig.show()
```
