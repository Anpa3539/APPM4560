## c) Simulating a game using interarrival times

```python
from functions import *

# Simulation of interarrival times
def game_score_single_team(l:float,T:float):
    """ 
    Find the score of a team throughout a game

    Args:
        l: Score frequency (1/minute)
        T: length of game (minutes)
    """

    t = 0 
    N = 0

    times = []

    while t < T:
        Ui = np.random.uniform(0,1)
        ti = -1/l * np.log(Ui)
        t += ti
        if t < T:
            N+=1
            times.append(t)

    score_vector = list(range(0,N))

    return {'score':N,'times':times,'score_vector':score_vector}

# Simulating games
l = 3 # /minute
T = 48 # minutes
A_game = game_score_single_team(l,T)
B_game = game_score_single_team(l,T)

# plotting
t_A = A_game['times']
t_B = B_game['times']
names = ['Team B','Team A']
plot_event_timeline(t_A,t_B,0,T,title='Simulated Game Score Times',xlabel='Time (minutes)',blue_label='Team B',red_label='Team A')
plot_2d([B_game['score_vector'],A_game['score_vector']],[t_B,t_A],line_name=names,title='Score vs Time',xlabel='Time (minute)',ylabel='Score')
```

### d) Simulating a game by assigning to either team w.p. 1/2

```python
from functions import *

def game_score_full_game(l,T):
    """ 
    Calculate the game outcome for both teams with equal probability of scoring

    Args:
        l: probability of one team scoring
        T: game length
    """

    # Finding true frequency of a basket being scored (2 teams at frequency l)
    l = 2*l  

    t = 0 
    N_A = 0
    N_B = 0
    t_A = []
    t_B = []

    teams = ['A','B']
    probabilities = [0.5,0.5]

    while t < T:
        Ui = np.random.uniform(0,1)
        ti = -1/l * np.log(Ui)
        t += ti
        if t < T:
            team = np.random.choice(teams,p=probabilities)
            if team == 'A':
                N_A += 1
                t_A.append(t)
            else: 
                N_B += 1
                t_B.append(t)

    score_vector_A = list(range(0,N_A))
    score_vector_B = list(range(0,N_B))

    return {'N_A':N_A,'t_A':t_A,'score_vector_A':score_vector_A,'N_B':N_B,'t_B':t_B,'score_vector_B':score_vector_B}


l = 3
T = 48
game = game_score_full_game(l,T)  
 
# plotting
t_A = game['t_A']
t_B = game['t_B']
names = ['Team B','Team A']
plot_event_timeline(t_A,t_B,0,T,title='Simulated Game Score Times',xlabel='Time (minutes)',blue_label='Team B',red_label='Team A')
plot_2d([game['score_vector_B'],game['score_vector_A']],[t_B,t_A],line_name=names,title='Score vs Time',xlabel='Time (minute)',ylabel='Score')
```

### e) Simulating $10^5$ games

```python
import numpy as np
from functions import *

def simulate_many_games(l,t,n=10**5):
    """ 
    Efficiently simuate n games between teams A and B

    Args:
        l: rate of each team of scoring
        t: time
        n: number of games to simulate
    """

    # Simulating scores in n games
    N = np.random.poisson(l*2*t,size=n)

    # Simulating A scores
    NA = np.random.binomial(N,0.5)

    # Finding the score difference
    D = 4*NA - 2*N

    # Expected score difference and variance
    E_D = np.mean(D)
    Var_D = np.var(D)

    # Probability of a tie
    P_tie = np.mean(np.abs(D) <1)

    return {'E_D':E_D,'Var_D':Var_D,'P_tie':P_tie}

simulate_many_games(3,48,10**5)

```
