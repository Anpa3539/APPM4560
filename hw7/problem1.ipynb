#### a) Find the probability that team A wins with $\lambda_A=2$ and $\lambda_B = 1.5$

#### b) Find the probability of a tie with respect to time

```python
import numpy as np
from functions import plot_2d,poisson

lA = 2/90   # scores/min
lB = 1.5/90 # scores/min

b = 14

t_list = np.linspace(0,90,100)

def prob_tie(t,lA,lB,b,T = 90):
    """ 
    Find the probability of a tie with respect to time
    """

    s = 0

    # Calculating time variant lambda
    lA_t = lA*(T-t)
    lB_t = lB*(T-t)

    for i in range(b):
        s += poisson(lA_t,i)*poisson(lB_t,i)

    return s

p_tie = [prob_tie(t,lA,lB,b) for t in t_list]

plot_2d(p_tie,x=t_list,xlabel='Time (minutes)', ylabel='Probability of a Tie',title='Probability of a Tie With Respect to Time')
```

#### c) Find probability of a tie given A scores once at $t=60$ minutes

```python
import numpy as np
from functions import poisson

lA = 2/90   # scores/min
lB = 1.5/90 # scores/min

b = 14

t_list = np.linspace(0,30,100)

def prob_tie(t,lA,lB,b,T = 90):
    """ 
    Find the probability of a tie with respect to time
    """

    s = 0

    # Calculating time variant lambda
    lA_t = lA*(T-t)/3
    lB_t = lB*(T-t)/3

    for i in range(b):
        s += poisson(lB_t,1)*poisson(lA_t,i)*poisson(lB_t,i)

    return s

p_tie = [prob_tie(t,lA,lB,b) for t in t_list]
t_adjusted = [t + 60 for t in t_list]
plot_2d(p_tie,x=t_adjusted,xlabel='Time (minutes)', ylabel='Probability of a Tie',title='Probability of a Tie With Respect to Time')
```
