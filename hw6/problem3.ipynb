### Problem 3
#### Calculate the probability of extinction

```python
import numpy as np
from scipy.optimize import root_scalar

l_list = [0.5,1.0,2.0]

for l in l_list:
    f = lambda x: np.exp(l*(x-1)) - x
    root_data = root_scalar(f,x0 = 0.5)
    print(f'lambda = {l}: P(extinction) = {round(root_data.root,5)}')
```
