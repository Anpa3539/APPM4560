#### Simulate markov chain

```python
import numpy as np
import random

a = 0.49

# Generation limit
limit = 200


# Simulating the avalanches
num_avalanche = int(1e3)
extinct_avalanches = 0
final_quantity = np.zeros(num_avalanche)
num_steps = np.zeros(num_avalanche)
avalanches = np.zeros([num_avalanche,limit+1])

for i in range(num_avalanche):
    n = 0
    Xn = 1
    process = np.zeros(limit + 1)
    process[n] = Xn
    while n < limit and Xn > 0:
        
        # Random step choice
        Zn = random.binomialvariate(Xn,1-a)
        Xn1 = 2*Zn

        # While loop stepping
        Xn = Xn1
        n+=1

        # Analytics
        if Xn == 0:
            extinct_avalanches +=1
        process[n] = Xn

    final_quantity[i] = Xn
    num_steps[i] = n
    avalanches[i] = process
    

print(f'{extinct_avalanches} avalanches died out in under 200 steps ({extinct_avalanches/num_avalanche*100} percent)')
    
import plotly.graph_objects as go

fig = go.Figure()
trace = go.Scatter(y=final_quantity,mode='lines',name='final')
fig.add_trace(trace)
fig.update_layout(title='Quantity of particles at Time 200',xaxis_title='Trial',yaxis_title='X_200')
fig.show()

fig = go.Figure()
trace = go.Scatter(y=num_steps,mode='lines',name='final')
fig.add_trace(trace)
fig.update_layout(title='Number of Steps Taken Before Extinciton or Limit',xaxis_title='Trial',yaxis_title='n')
fig.show()

fig = go.Figure()
for i in range(limit + 1):
    trace = go.Scatter(y=avalanches[i],mode='lines',name='')
    fig.add_trace(trace)
fig.update_layout(title='Number of Particles vs Time',xaxis_title='Time',yaxis_title='Number of Particles',showlegend=False)
fig.show()




        
```

##### Analytical probabilty of extinction

```python
rho = lambda a: (1 - np.sqrt(4*a**2 - 4*a + 1))/ (2 - 2*a)
print(f'rho = {round(rho(a),4)}')
```
