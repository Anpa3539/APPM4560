### Problem 1: Grid Markov Chain
#### b: Find the stationary distribution

```python
import numpy as np

p = np.array([[0,1,0,0,0],
              [1/3,0,2/3,0,0],
              [0,3/4,0,1/4,0],
              [0,0,2/3,0,1/3],
              [0,0,0,1,0]])

evals, evecs = np.linalg.eig(p.T)
evecs = [evecs[:,i] for i in range(len(evecs))]
pi = evecs[3]/sum(evecs[3])
print(f'pi = {pi}')
print(f'lambda_i = {evals}')
```

#### c: Finding $q_{50}$

```python
p50 = np.linalg.matrix_power(p,50)
print(f'p^50 = {p50}')
q50 = p50@np.array([0,0,1,0,0])
q50 = q50/sum(q50)
print(f'q50 = {q50}')
```

#### d: Plotting $q_{50}$ and $\vec{\pi}$

```python
import plotly.graph_objects as go

fig = go.Figure()
q_trace = go.Scatter(y=q50,mode='lines',name='q50')
pi_trace = go.Scatter(y=pi,mode='lines',name='pi')
fig.add_trace(q_trace)
fig.add_trace(pi_trace)
fig.update_layout(title='Stationary Distribution vs q50 Approximation',xaxis_title='i',yaxis_title='pi(i)')
fig.show()
```
